### Імпорт та перевірка версії

```python
import requests
print(f"Requests version: {requests.__version__}")
```

### Використання різних методів (завдання з документації)

```python
# Тестуємо різні HTTP методи
base_url = 'https://httpbin.org'

# GET запит з параметрами
r_get = requests.get(f"{base_url}/get", params={"user": "Nazar", "task": "lab6"})
print(f"GET status: {r_get.status_code}")

# POST запит (відправка даних)
r_post = requests.post(f"{base_url}/post", json={"message": "Hello from Jupyter!"})
print(f"POST response: {r_post.status_code}")

# DELETE запит
r_delete = requests.delete(f"{base_url}/delete")
print(f"DELETE status: {r_delete.status_code}")
```

### Робота з заголовками та контентом

```python
r = requests.get('https://api.github.com')

print("Тип контенту:", r.headers['Content-Type'])
print("Текст відповіді (перші 100 символів):", r.text[:100])
```
